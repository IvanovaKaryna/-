### Завдання 33
Нормалізуйте **кожен стовпець** матриці:

```text
A = [[10, 20, 30],
     [20, 30, 40],
     [30, 40, 50]]
```

за формулою:

\[
z = \frac{x - \mu}{\sigma}
\]

де `μ` — середнє значення стовпця, `σ` — стандартне відхилення стовпця.

Не використовуйте цикл `for`.

````python
# Код студента
A = np.array([
    [10, 20, 30],
    [20, 30, 40],
    [30, 40, 50]
], dtype=float)

mean = np.mean(A, axis=0)
std = np.std(A, axis=0)
Z = (A - mean) / std
print("Початкова матриця:")
print(A)
print("\nСереднє кожного стовпця:")
print(mean)
print("\nСтандартне відхилення кожного стовпця:")
print(std)
print("\nНормалізована матриця:")
print(Z)
````

```text
sales = [[120, 150, 170, 160],

         [200, 180, 210, 230]]
```

Рядки — товари, стовпці — місяці.

Визначте:
1. загальний обсяг продажів кожного товару;

Створіть квадратну матрицю `5 × 5` із випадкових цілих чисел від `1` до `20`.

Виконайте:
1. знайдіть суму головної діагоналі;
2. знайдіть суму побічної діагоналі;
3. знайдіть максимальний елемент кожного рядка;
4. знайдіть мінімальний елемент кожного стовпця;
5. замініть усі елементи, менші за середнє значення всієї матриці, на `0`.

````python
# Код студента
np.random.seed(42)
matrix = np.random.randint(1, 21, (5, 5))
main_diagonal_sum = np.trace(matrix)
secondary_diagonal_sum = np.trace(np.fliplr(matrix))
row_max = np.max(matrix, axis=1)
column_min = np.min(matrix, axis=0)
mean_value = np.mean(matrix)
result = matrix.copy()
result[result < mean_value] = 0

print("Початкова матриця:")
print(matrix)
print("\nСереднє значення матриці:", mean_value)
print("\nСума головної діагоналі:", main_diagonal_sum)
print("Сума побічної діагоналі:", secondary_diagonal_sum)
print("\nМаксимальний елемент кожного рядка:")
print(row_max)
print("\nМінімальний елемент кожного стовпця:")
print(column_min)
print("\nМатриця після заміни елементів, менших за середнє, на 0:")
print(result)
````

Створіть матрицю `4 × 4` із довільних цілих чисел.

Для кожного рядка знайдіть його норму та створіть нову матрицю, у якій кожен рядок нормалізований до одиничної довжини.

### Завдання 40
Створіть матрицю коефіцієнтів і вектор правих частин для системи:

```text
 x + 2y +  z =  9
2x +  y -  z =  8
3x -  y + 2z =  3
```

1. Розв’яжіть систему за допомогою `np.linalg.solve()`.
2. Перевірте знайдений розв’язок матричним множенням `A @ x`.
3. Порівняйте результат із вектором `b`.

````python
# Код студента
A = np.array([
    [1, 2, 1],
    [2, 1, -1],
    [3, -1, 2]
], dtype=float)

b = np.array([9, 8, 3], dtype=float)
x = np.linalg.solve(A, b)
result = A @ x

print("Матриця коефіцієнтів A:")
print(A)
print("\nВектор правих частин b:")
print(b)
print("\nРозв'язок системи:")
print(x)
print("\nA @ x:")
print(result)
print("\nВектор b:")
print(b)
print("\nЧи дорівнює A @ x вектору b:",
      np.allclose(result, b))
````
